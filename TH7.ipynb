```python
#W7A1
def binary_search(arr, left, right, target):
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] > target:
            return binary_search(arr, left, mid - 1, target)
        else:
            return binary_search(arr, mid + 1, right, target)
    return -1
if __name__=="__main__":
    arr = list(map(int, input().split()))
    target = int(input())
    print(binary_search(arr, 0, len(arr)-1, target))
```

```python
#W7A1
def binary_search(arr, value):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == value:
            return mid
        elif arr[mid] < value:
            left = mid + 1
        else:
            right = mid -1
    return -1
```

```python
#W7A3
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[min_idx] > arr[j]:
                min_idx = j
        arr[min_idx], arr[i] = arr[i], arr[min_idx]
    return arr
if __name__=="__main__":
    arr = list(map(int, input().split()))
    print(bubble_sort(arr))
    print(selection_sort(arr))
```

```python
#W7A7
def day_tang_nghiem_ngat(N):
    do_dai = -1
    n = len(N)
    for i in range(n):
        N_1 = []
        N_1.append(N[i])
        for j in range(i+1, n-1):
            if N[i] < N[j]:
                N_1.append(N[j])
                i = j
        if do_dai < len(N_1):
            do_dai = len(N_1)
    return do_dai
N = [7, 7, 7, 7, 7, 7, 7]
T = (day_tang_nghiem_ngat(N))
print(T)
```

```python
#W7A8
def do_dai_min(interval, queries):
    d_1 = []
    for i in queries:
        found = False
        minn = float('inf')
        for j,k in interval:    
            if j <= i <= k:
                do_dai = k - j + 1
                if do_dai < minn:
                    minn = do_dai
                    found = True
        if found == True:
            d_1.append(minn)
        else:
            d_1.append(-1)
    return d_1
interval = [[1, 4], [2, 4], [3, 6], [4, 4]]
queries = [2, 3, 4, 5] 
print(do_dai_min(interval, queries))
```

```python
#W7A9
def nho_hon(num):
    d = []
    for i in range(len(num)):
        dem = 0
        for j in range(i+1, len(num)):
            if num[i] > num[j]:
                dem += 1
        d.append(dem)
    return(d)
num = [5, 4, 3, 2, 1] 
print(nho_hon(num))
```

```python
#W7A10
def chia_du(nums, x):
    n = len(nums)
    d = []
    maxx = 0
    for i in range(n):
        tong = nums[i]
        for j in range(i+1, n):
            tong += nums[j]
            du = tong % x
            if du > maxx:
                maxx = du
        d.append(maxx)
    return max(d)
nums = [1, 5, 9, 13, 17] 
x = 10
print(chia_du(nums, x))
```

```python
#W7A13
def day_hai_hoa(nums):
    n = len(nums)
    day_ = sorted(set(nums))
    dem = 0
    for i in range(len(day_)-1):
        count = 0
        if day_[i+1] - day_[i] == 1:
            for j in nums:
                if j == day_[i] or j == day_[i+1]:
                    count += 1
            if dem < count:
                dem = count
    return dem
if __name__ == "__main__":
    nums = list(map(int, input().split()))
    print(day_hai_hoa(nums))
```

```python
#W7A14
def vi_tri(nums, n):
    d = []
    for i in range(n):
        if nums[i] == 7:
            d.append(i)
    if d == []:
        return "Not found"
    return d[::-1]
n = int(input())
nums = list(map(int, input().split()))
print(vi_tri(nums, n))
```
