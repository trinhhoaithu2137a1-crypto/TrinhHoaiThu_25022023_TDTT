```python
#9A1
def merge(a, b):
    i = j = 0
    res = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            res.append(a[i])
            i += 1
        else:
            res.append(b[j])
            j += 1
    res.extend(a[i:])
    res.extend(b[j:])
    return res
```

```python
#9A2:
def kth_largest(mat, k):
    arr = []
    for row in mat:
        arr.extend(row)
    arr.sort(reverse=True)
    return arr[k-1]
```

```python
#9A3:
def min_after_one_swap(s):
    best = int(s)
    n = len(s)
    a = list(s)

    for i in range(n):
        for j in range(i+1, n):
            a[i], a[j] = a[j], a[i]
            if a[0] != '0':
                best = min(best, int("".join(a)))
            a[i], a[j] = a[j], a[i]
    return best
```

```python
#9A4:
def can_convert(a, b):
    n, m = len(a), len(b)
    dp = [[False]*(m+1) for _ in range(n+1)]
    dp[0][0] = True

    for i in range(n):
        for j in range(m+1):
            if not dp[i][j]:
                continue
            # xóa ký tự thường
            if a[i].islower():
                dp[i+1][j] = True
            # chuyển thành hoa nếu khớp
            if j < m and a[i].upper() == b[j]:
                dp[i+1][j+1] = True

    return dp[n][m]
```

```python
#9A5:
def is_subarray(a, b):
    n, m = len(a), len(b)
    for i in range(m - n + 1):
        if b[i:i+n] == a:
            return True
    return False
```

```python
#9A6:
def next_permutation(a):
    n = len(a)
    i = n - 2
    while i >= 0 and a[i] >= a[i+1]:
        i -= 1
    if i < 0:
        return a[::-1]

    j = n - 1
    while a[j] <= a[i]:
        j -= 1
    a[i], a[j] = a[j], a[i]
    a[i+1:] = reversed(a[i+1:])
    return a
```

```python
#9A7:
def canPlaceFlowers(flowerbed, k):
    n = len(flowerbed)
    i = 0
    while i < n and k > 0:
        if flowerbed[i] == 0:
            left = (i == 0 or flowerbed[i-1] == 0)
            right = (i == n-1 or flowerbed[i+1] == 0)
            if left and right:
                flowerbed[i] = 1
                k -= 1
        i += 1
    return k == 0
```

```python
#9A8:
def is_path_crossing(moves):
    x = y = 0
    visited = {(0,0)}

    for m in moves:
        if m == 'R': x += 1
        elif m == 'L': x -= 1
        elif m == 'U': y += 1
        elif m == 'D': y -= 1

        if (x, y) in visited:
            return True
        visited.add((x, y))
    return False
```

```python
#9A9:
def Median(nums):
    a = nums[:]
    # sắp xếp thủ công nhỏ
    for i in range(5):
        for j in range(i+1, 5):
            if a[i] > a[j]:
                a[i], a[j] = a[j], a[i]
    return a[2]
```

```python
#9A10:
import bisect

def findRadius(houses, heaters):
    houses.sort()
    heaters.sort()
    ans = 0

    for h in houses:
        idx = bisect.bisect_left(heaters, h)
        d1 = abs(h - heaters[idx]) if idx < len(heaters) else float('inf')
        d2 = abs(h - heaters[idx-1]) if idx > 0 else float('inf')
        ans = max(ans, min(d1, d2))

    return ans
```
