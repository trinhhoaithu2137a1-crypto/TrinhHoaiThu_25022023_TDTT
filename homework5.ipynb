```python
#W6A19
def max_(songuyen):
    n = len(songuyen)
    max_ = -3*10**18
    for i in range(n):
        if songuyen[i] > max_:
            max_ = songuyen[i]
    return max_
if __name__ == "__main__":
    songuyen = list(map(int, input().split()))
    print(max_(songuyen))
```

```python
#W6A20
def index(so_nguyen, k):
    n = len(so_nguyen)
    for i in range(n):
        if so_nguyen[i] == k:
            return i
    return -1
if __name__=="__main__":
    so_nguyen = list(map(int, input().split()))
    k = int(input())
    print(index(so_nguyen, k))
```

```python
#W6A21
data = input().split()
n = len(data)
d = {}
for x in range(n):
    word = data[x]
    if not word in d:
        d[word] = []
    d[word].append(x)
for key in d:
    d[key] = tuple(d[key])
print(d)
```

```python
#W6A22
def max_(songuyen):
    n = len(songuyen)
    max_ = -3*10**18
    index = 0
    for i in range(n):
        if songuyen[i] > max_:
            max_ = songuyen[i]
            index = i + 1
    return index
if __name__ == "__main__":
    songuyen = list(map(int, input().split()))
    print(max_(songuyen))
```

```python
#W6A23
def xoa(so_nguyen, x):
    l = []
    for i in so_nguyen:
        if i >= x:
            l.append(i)
    return l
if __name__ == "__main__":
    so_nguyen = list(map(int, input().split()))
    x = int(input())
    print(xoa(so_nguyen, x))
```

```python
#W6A24
def zero(so):
    zero_ = []
    khac = []
    for i in so:
        if i == 0:
            zero_.append(i)
        else:
            khac.append(i)
    return zero_ + khac
if __name__ == "__main__":
    so = list(map(int, input().split()))
    print(*zero(so))
```

```python
#W6A26
n = list(map(int, input().split()))
do_dai = len(n)
tong = 0
for i in range(1, do_dai-1):
    if n[i] > n[i-1] and n[i] > n[i+1]:
        tong += n[i]
print(tong)
```

```python
#W6A27
def be_hon_x(so_nguyen, x):
    n = len(so_nguyen)
    d = []
    for i in range(n):
        if so_nguyen[i] < x:
            d.append(i)
    return -1
if __name__ == "__main__":
    so_nguyen = list(map(int, input().split()))
    x = int(input())
    print(be_hon_x(so_nguyen, x))
```
