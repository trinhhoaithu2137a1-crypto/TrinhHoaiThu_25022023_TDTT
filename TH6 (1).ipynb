```python
#W6A4
data = input().split()
res = {}
for item in data:
    key, value = item.split(':') # dùng .split(':') để tách riêng key và value
    if key in res:
        res[key].append(value)
    else:
        res[key] = [value]
print(res)
```

```python
#W6A5
so_nguyen = list(map(int, input().split()))
danh_sach = {'positives' : 0, 'negatives' : 0, 'zeros' : 0}
for num in so_nguyen:
    if num > 0:
        danh_sach['positives'] += 1
    elif num < 0:
        danh_sach['negatives'] += 1
    else:
        danh_sach['zeros'] += 1
print(danh_sach)
```

```python
#W6A9
def nhap_dict():
    d = {}
    danhsach = input().split()
    for item in danhsach:
        key, value = item.split(':')
        d[key] = int(value)
    return d
if __name__ == "__main__":
    dict_1 = nhap_dict()
    dict_2 = nhap_dict()
    res = dict_1.copy()
    for key, value in dict_2.items():
        if key in res:
            res[key] += value
        else:
            res[key] = value
    res_1 = sorted(res.keys())
    ketqua = {}
    for x in res_1:
        ketqua[x] = res[x]
    print(ketqua)
```

```python
#W6A10
danh_sach = list(map(int, input().split()))
k = int(input())
n = len(danh_sach)
map_1 = []
for i in range(n):
    for j in range(i+1, n):
        if danh_sach[i] + danh_sach[j] == k:
            map_1.append((danh_sach[i], danh_sach[j]))
print(map_1)
```

```python
#W6A13
def nhap_dict():
    d = {}
    danh_sach = input().split()
    for item in danh_sach:
        key, value = item.split(':')
        d[key] = value
    return d
if __name__ == '__main__':
    dict_1 = nhap_dict()
    dict_2 = {}
    for key, value in dict_1.items():
        dict_2[value] = key
    print(dict_2)
#dict_2 = {value: key for key, value in dict_1.items()}
```

```python
#W6A14
dsach1 = list(map(int, input().split()))
dsach2 = list(map(int, input().split()))
nums = [x for x in dsach1 if x in dsach2]
map_ = []
for i in nums:
    if not i in map_:
        map_.append(i)
print(map_)
```

```python
#W6A15
data = input().split()
dict_1 = {}
for item in data:
    key, value = item.split(':')
    dict_1[key] = value
k = int(input())
d = {}
for key, value in dict_1.items():
    if int(value) > k:
        d[key] = value
print(d)
```
